# درک کسب‌وکار (Business Understanding)
### شکار مشتریان در آستانه‌ی ریزش، فروشگاه آنلاین (Online Retail، دسامبر ۲۰۱۰ تا دسامبر ۲۰۱۱)

</div>

<div dir="rtl" style="text-align:right; font-family:Vazirmatn,Tahoma,'Segoe UI',sans-serif; line-height:2.1; font-size:16px">

## ۱. سناریو

یک فروشگاه اینترنتی بین‌المللی تعدادی از مشتریان خود را به‌تدریج از دست می‌دهد. در فروشگاه آنلاین مشتری هیچ‌وقت «اعلام ترک» نمی‌کند و تنها نشانه‌ی رفتن او این است که دیگر خرید نمی‌کند.

داده‌ها تراکنش‌های یک‌ساله‌ی فروشگاه از دسامبر ۲۰۱۰ تا دسامبر ۲۰۱۱ است و **هیچ ستونی با عنوان «ریزش» در آن وجود ندارد**؛ پس ریزش باید از روی رفتار خرید مشتریان تعریف و ساخته شود.

شرکت می‌خواهد با بودجه‌ی محدود **۵۰۰ تماس**، مشتریان در معرض ریزش را بشناسد و برای نگه‌داشتن آن‌ها اقدام کند.

</div>

<div dir="rtl" style="text-align:right; font-family:Vazirmatn,Tahoma,'Segoe UI',sans-serif; line-height:2.1; font-size:16px">

## ۲. مسئله‌ی کسب‌وکار چیست؟

مشتریان فروشگاه در حال ریزش هستند و شرکت نمی‌داند کدام مشتریان در خطرند. جذب مشتری جدید معمولاً گران‌تر از نگه‌داشتن مشتری فعلی است، پس هر مشتری که بی‌صدا از دست برود مستقیماً به درآمد آینده آسیب می‌زند.

مسئله دو لایه دارد:

- **شناسایی:** مشتریان در آستانه‌ی ریزش چه کسانی‌اند و ریزش در این کسب‌وکار چه شکلی دارد؟ (برچسب ریزش وجود ندارد.)
- **اقدام:** با فقط ۵۰۰ تماس، تماس با کدام مشتریان بیشترین فایده را دارد؟

</div>

<div dir="rtl" style="text-align:right; font-family:Vazirmatn,Tahoma,'Segoe UI',sans-serif; line-height:2.1; font-size:16px">

## ۳. هدف کسب‌وکار (در یک خط)

> با بودجه‌ی حداکثر ۵۰۰ تماس، مشتریان در معرض ریزش را شناسایی کنیم تا بتوان با اقدام به‌موقع، آن‌ها را حفظ کرد.

</div>

<div dir="rtl" style="text-align:right; font-family:Vazirmatn,Tahoma,'Segoe UI',sans-serif; line-height:2.1; font-size:16px">

## ۴. محدودیت‌های پروژه چیست؟

- **نبود برچسب ریزش:** ستون «ریزش» در داده نیست و باید از رفتار خرید ساخته شود؛ پس بخشی از کار، تعریف و ساخت خود هدف (برچسب) است.
- **بودجه‌ی ۵۰۰ تماس:** تعداد مشتریان پرخطر احتمالاً بیشتر از ۵۰۰ نفر است؛ پس خروجی باید یک لیست **رتبه‌بندی‌شده** باشد، نه فقط برچسب «ریزشی / غیرریزشی».
- **داده‌ی ثابت:** داده‌ی جدیدی اضافه نمی‌شود و کار فقط روی همین داده انجام می‌شود. پروژه یک تحلیل روی یک «عکس ثابت» از کسب‌وکار است، نه یک سیستم زنده.
- **بازه‌ی یک‌ساله:** فقط حدود ۱۲ ماه داده داریم؛ اثر فصلی (مثلاً اوج خرید پایان سال) را نمی‌توان کامل سنجید و تعمیم‌پذیری نتایج محدود است.
- **نیاز به تاریخ مرجع:** برچسب با نگاه به ۹۰ روز بعد از یک «تاریخ مرجع» ساخته می‌شود؛ پس تاریخ مرجع باید حدود ۹۰ روز قبل از پایان داده باشد و ویژگی‌ها فقط از قبل از آن ساخته شوند (جلوگیری از نشت اطلاعات).
- **تعریف ریزش یک قرارداد تحلیلی است:** «۹۰ روز بدون خرید» به معنای لغو قطعی نیست و ممکن است مشتری بعداً برگردد.
- **نبود سنجش اثر تماس:** تماسی واقعاً انجام نمی‌شود؛ پس معیار موفقیت «دقت انتخاب ۵۰۰ نفر» است، نه «تعداد مشتریان نجات‌یافته».

</div>

<div dir="rtl" style="text-align:right; font-family:Vazirmatn,Tahoma,'Segoe UI',sans-serif; line-height:2.1; font-size:16px">

## ۵. هدف داده‌کاوی چیست؟

ریزش در این فروشگاه یک رویداد ثبت‌شده نیست و فقط از الگوی رفتار مشتری فهمیده می‌شود. به همین دلیل هدف داده‌کاوی در دو گام تعریف می‌شود:

- **گام اول، کشف رفتار:** مشتریان را بر اساس ویژگی‌های رفتاری قبل از تاریخ مرجع (مانند Recency، Frequency و Monetary) **خوشه‌بندی** می‌کنیم تا انواع مشتری و گروه‌های دارای نشانه‌ی ریزش را بشناسیم.
- **گام دوم، شناسایی ۵۰۰ نفر:** برچسب ریزش را با قاعده‌ی «بدون خرید معتبر در ۹۰ روز پس از تاریخ مرجع» می‌سازیم، یک مدل **طبقه‌بندی** آموزش می‌دهیم و مشتریان را بر اساس احتمال ریزش رتبه‌بندی می‌کنیم.

**معیار موفقیت:** از ۵۰۰ نفر انتخاب‌شده چند نفر واقعاً ریزش کرده‌اند (Precision@500)، و مقایسه‌ی آن با انتخاب تصادفی و با یک مدل ساده‌ی پایه مبتنی بر RFM.

</div>

<div dir="rtl" style="text-align:right; font-family:Vazirmatn,Tahoma,'Segoe UI',sans-serif; line-height:2.1; font-size:16px">

## ۶. توضیح برای دو بخشی بودن سناریو

سناریو دو بخش دارد چون دو پرسش متفاوت را باید جواب بدهد.

**بخش اول: خوشه‌بندی (Clustering)**

- پرسش: مشتریان چه الگوهای رفتاری دارند و ریزش در این کسب‌وکار چه شکلی است؟
- نوع یادگیری: بدون ناظر (Unsupervised)، برچسب لازم ندارد.
- چرا لازم است: چون برچسب ریزش نداریم، اول باید رفتار مشتریان را بفهمیم تا تعریف ریزش را درک و اعتبارسنجی کنیم.
- خروجی: گروه‌های رفتاری و درک بهتر از مشتریان.

**بخش دوم: طبقه‌بندی (Classification)**

- پرسش: کدام ۵۰۰ مشتری بیشترین احتمال ریزش را دارند؟
- نوع یادگیری: با ناظر (Supervised)، به برچسب ریزش نیاز دارد.
- چرا لازم است: چون در نهایت باید احتمال ریزش را پیش‌بینی و مشتریان را رتبه‌بندی کنیم.
- خروجی: احتمال ریزش هر مشتری و لیست ۵۰۰ نفره.

**نکته‌ی مهم:** خوشه‌بندی باید فقط روی داده‌ی **قبل از تاریخ مرجع** انجام شود. اگر خوشه‌بندی روی داده‌ی ۹۰ روز آینده هم انجام شود و شماره‌ی خوشه به مدل داده شود، اطلاعات آینده به مدل نشت می‌کند و نتیجه جعلی خواهد شد.

</div>

## ۷. هدف داده‌کاوی (در یک خط)

> ابتدا با خوشه‌بندی الگوهای رفتاری مشتریان را کشف کنیم، سپس با یک مدل طبقه‌بندی که برچسبش از قاعده‌ی ۹۰ روز بدون خرید ساخته شده، ۵۰۰ مشتری با بیشترین احتمال ریزش را رتبه‌بندی و انتخاب کنیم.

</div>

## ۸. خروجی نهایی

خروجی پایانی یک دیکشنری از ۵۰۰ مشتری پرخطر است. کلید آن `CustomerID` و مقدار آن اطلاعات همان مشتری است، و در پایان به‌صورت جدول نمایش داده می‌شود.
